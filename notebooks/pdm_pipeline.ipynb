```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from src.data_loader import load_raw_dataset
from src.features import extract_features_from_signal, compute_rms_ratio
from src.features_engine import convert_to_operating_time  # 🔹 ДОБАВЛЕН ИМПОРТ
from src.degradation import detect_fpt
from src.models import train_rf_model, predict_rul_exponential, save_model
from src.visualization import plot_rul_comparison, plot_feature_importance
from src.config import DATA_DIR

# 1. Загрузка сырых данных
df_raw = load_raw_dataset("2nd_test")

# 2. Извлечение базовых признаков (RMS, Kurtosis, Skew и т.д.)
# Функция должна вернуть DataFrame с колонками вида b1_rms, b2_rms...
print("⚙️ Извлечение признаков из окон...")
df = extract_features_batch(df_raw)

# 3. Расчет медианы RMS (требуется для фильтрации простоев в convert_to_operating_time)
rms_cols = [c for c in df.columns if c.endswith('_rms')]
df['rms_median'] = df[rms_cols].median(axis=1)

# 4. Пересчет календарного времени в моточасы (теперь колонка rms_median существует)
df = convert_to_operating_time(df)

# 5. Расчет относительных признаков (RMS Ratio)
df = compute_rms_ratio(df)

# 6. Детекция точки начала деградации (FPT)
bearing = "b3"
fpt = detect_fpt(df, bearing)
print(f"🔍 FPT для {bearing}: {fpt:.2f} ч")

# 7. Обучение ML-модели и визуализация
rf, feats = train_rf_model(df, bearing)
plot_rul_comparison(df, bearing, rf, feats)
plot_feature_importance(rf, feats, bearing)
save_model(rf, feats, bearing)

# 8. Экспоненциальный прогноз RUL
exp_res = predict_rul_exponential(df, bearing, fpt)
print("🔮 Экспонента:", exp_res)
```
